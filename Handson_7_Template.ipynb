## 8.1 Tip Calculator
Add exception handling to a Tip Calculator program.

### Console:
```powershell
Tip Calculator

INPUT
Cost of meal: ten
Must be valid decimal number. Please try again. 
Cost of meal: -10
Must be greater than 0. Please try again. 
Cost of meal: 52.31
Tip percent:	17.5
Must be valid integer. Please try again. 
Tip percent:	20

OUTPUT
Cost of meal:   52.31 
Tip percent:    20%
Tip amount:     10.46
Total amount:   62.77
```

### Specifications:
- The program should accept **decimal** entries like 52.31 and 15.5 for the cost of the meal.
- The program should accept **integer** entries like 15, 20, 25 for the tip percent.
- The program should validate both user entries. That way, the user can’t crash the program by entering invalid data.
- The program should only accept numbers that are **greater than 0**.
- The program should round results to a maximum of two decimal places.


````python
def get_com():
    while True:
        try:
            com=float(input("Cost of meal: "))
            if com < 0:
                print("Must be greater than 0. Please try again.")
            else:
                return com
        except ValueError:
            print("Must be valid decimal Number. Please try again.")
        
def get_tip():
    while True:
        try:
            tip = int(input("Tip percent: "))
            return tip
        except ValueError:
            print("Must be valid integer. Please try again.")
def main():
    print("Tip Calculator")
    com = get_com()
    tip = get_tip()
    tipamount= round((com*(tip/100)),2)
    total = round((com+tipamount),2)
    print("\nOUTPUT")
    print("Cost of meat:  ",com)
    print("Tip percent:", tip,"%")
    print("Tip amount:", tipamount)
    print("Total amount:",total,"\n")
main()
````

### Specifications:

- Handle all exceptions that might occur so that the user can’t cause the program to crash. If the all items file is missing, display an appropriate error message and exit the program.
- If the inventory file is missing, display an appropriate error message and continue with an empty inventory for the user. That way, the program will write a new inventory file when the user adds items to the inventory.

````python
import random
import sys
def write_wiz(wizitems):
        with open("wizard_inventory.txt",'w',newline="")as file:
            for item in wizitems:
                file.write(item )
    
def read_wiz():
    wizitems = []
    try:        
        with open("wizard_inventory.txt",'r',newline="")as file:
            for line in file:
                wizitems.append(line) 
    except FileNotFoundError:
        print("\nCould not find inventory file!")
        print("Wizard is starting with no inventory.")
    return wizitems
   
def walk(wizitems):
    wiz = []
    try:
        with open("wizard_all_items.txt",'r')as file:
            for line in file:
                wiz.append(line)
    except:
        print("Couldn't find items file ")
        print("Exiting program. Bye!")
        return False
        
          
    x= random.choice(wiz)
    print("While walking down a path, you see ",x)
    a=input("Do you want to grab it?(y/n): ")
    if a=='y':
        if len(wizitems)>=4:
            print("You cant carry more items. Drop something first.")
        else:
            wizitems.append(x)
            write_wiz(wizitems)
            print("You picked up ",x)
    elif a=='n':
        print("You refused to pick up ",x)
    else:
        print("Enter correct command")
    
            
def show(wizitems):
    try:
        for i in range (len(wizitems)):
            item = wizitems[i]
            print(str(i+1)+ ". "+item)
    except:
        print("Cannot display non-existant file")
        return
def drop(wizitems):
    try:
        number = int(input("Number: "))
        item = wizitems.pop(number-1)
        write_wiz(wizitems)
        print("You dropped "+ item)
    except:
        print("Invalid item number")
    
def main():
    print("The Wizard Inventory Program\n")
    print("COMMAND MENU")
    print("walk-Walk down the path")
    print("show-Show all items")
    print("drop-Drop an item")
    print("exit-Exit program")
    wizitems = read_wiz()
    x = " "
    while x != "exit":
        x= input("\nCommand: " )
        if x == "walk":
            a=walk(wizitems)
            if a==False:
                break
        elif x == "show":
            show(wizitems)
        elif x=="drop":
            drop(wizitems)
        elif x =="exit":
            print("Bye!")
        else:
            print("Please enter valid command.")
        
        
main()       
````
